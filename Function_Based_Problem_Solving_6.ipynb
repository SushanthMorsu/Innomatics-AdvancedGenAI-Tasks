Problem Statement 1:Smart Parking Lot Management System

```python
def parking_system(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)

    if parked > capacity:
        print("Parking Status: Full / Over Capacity")
    else:
        print("Parking Status: Available")

    print("Peak Parking Usage:", peak)

# Sample Input
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
parking_system(capacity, logs)
```

Online Food Delivery Time Estimator

```python
def delivery_time(distance, traffic, weather):
    time = distance * 5

    if traffic.lower() == "high":
        time += 10
    elif traffic.lower() == "medium":
        time += 5

    if weather.lower() == "rainy":
        time += 5
    elif weather.lower() == "storm":
        time += 15

    print("Estimated Delivery Time:", time, "minutes")


#Sample Input
distance = 8
traffic = "high"
weather = "rainy"
delivery_time(distance, traffic, weather)
```

Movie Theatre Seat Occupancy Analyzer

```python
def theatre_status(total_seats, booked):
    booked_count = len(booked)
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", occupancy, "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")


#Sample Input
total_seats = 200
booked = [1] * 150
theatre_status(total_seats, booked)
```

Cloud Server Load Classification System

```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    print("Average CPU Load:", avg, "%")

    if avg < 50:
        print("Server Status: Normal")
    elif avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]
server_load(cpu_readings)
```

Smart Classroom Resource Usage Monitor

```python
def resource_monitor(resources):
    overused = []

    for item, hours in resources.items():
        if hours > 8:
            overused.append(item)

    print("Overused Resources:", ", ".join(overused))

    if overused:
        print("Energy Alert: Yes")
    else:
        print("Energy Alert: No")


#Sample Input
resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}
resource_monitor(resources)
```

Online Event Registration Capacity Controller

```python
def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlist = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)


#Sample Input
capacity = 100
registrations = 105
event_registration(capacity, registrations)
```
