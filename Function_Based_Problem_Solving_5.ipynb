# Real-World Function-Based Problem Solving

## Problem 1: Social Media – Post Engagement Analyzer

```python
def analyze_post_engagement(likes):
    total_likes = sum(likes)
    if total_likes >= 1000:
        post_status = "Viral Post"
    else:
        post_status = "Normal Engagement"

    print(f"Total Likes: {total_likes}")
    print(f"Post Status: {post_status}")

# Demo
post_likes = [200, 300, 550]
analyze_post_engagement(post_likes)
```

## Problem 2: Healthcare – Medicine Stock Alert System

```python
def medicine_stock_alert(stock_level):
    if stock_level < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"

    print(f"Medicine Stock: {stock_level}")
    print(f"Status: {status}")

# Demo
medicine_stock_alert(6)
```

## Problem 3: Agriculture – Rainfall Adequacy Checker

```python
def rainfall_adequacy_checker(rainfall_data, required_level):
    total_rainfall = sum(rainfall_data)
    average_rainfall = total_rainfall / len(rainfall_data)

    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print(f"Average Rainfall: {int(average_rainfall)}")
    print(f"Rainfall Status: {status}")

# Demo
seasonal_rainfall = [60, 80, 75, 73]
required_rainfall = 70
rainfall_adequacy_checker(seasonal_rainfall, required_rainfall)
```

## Problem 4: Social Media – Duplicate Account Detection

```python
def detect_duplicate_accounts(usernames):
    unique_usernames = set()
    duplicates_found = False
    for username in usernames:
        if username in unique_usernames:
            duplicates_found = True
            break
        else:
            unique_usernames.add(username)

    if duplicates_found:
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

# Demo
account_list = ["user1", "user2", "user3", "user1"]
detect_duplicate_accounts(account_list)
```

## Problem 5: Healthcare – Appointment Eligibility Checker

```python
def check_appointment_eligibility(age):
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print(f"Patient Age: {age}")
    print(f"Eligibility Status: {status}")

# Demo
patient_age = 21
check_appointment_eligibility(patient_age)
```

## Problem 6: Agriculture – Premium Crop Price Filter

```python
def filter_premium_crops(crop_prices, premium_threshold=2000):
    premium_crops = [price for price in crop_prices if price > premium_threshold]
    print(f"Premium Crops: {premium_crops}")

# Demo
prices = [1800, 2500, 1500, 3200, 1900]
filter_premium_crops(prices)
```

## Problem 7: System Monitoring – Application Health Checker

```python
def check_application_health(error_count):
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print(f"Error Count: {error_count}")
    print(f"System Status: {status}")

# Demo
check_application_health(7)
```

## Problem 8: Banking – Daily Transaction Limit Checker

```python
def check_transaction_limit(transaction_amount, daily_limit=50000):
    if transaction_amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"

    print(f"Transaction Amount: {transaction_amount}")
    print(f"Transaction Status: {status}")

# Demo
check_transaction_limit(60000)
```

## Problem 9: E-Learning – Student Attendance Eligibility System

```python
def check_attendance_eligibility(attended_classes, total_classes):
    attendance_percentage = (attended_classes / total_classes) * 100

    if attendance_percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"

    print(f"Attendance Percentage: {attendance_percentage:.1f}")
    print(f"Exam Eligibility: {eligibility}")

# Demo
attended = 80
total = 100
check_attendance_eligibility(attended, total)
```

## Problem 10: Smart Electricity Bill Analyzer

```python
def calculate_electricity_bill(units_consumed):
    bill = 0

    if units_consumed <= 100:
        bill = units_consumed * 3
    elif units_consumed <= 200:
        bill = (100 * 3) + ((units_consumed - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units_consumed - 200) * 7)

    if bill < 500:
        usage_classification = "Low Usage"
    elif bill <= 1500:
        usage_classification = "Moderate Usage"
    else:
        usage_classification = "High Usage"

    print(f"Total Bill: ₹{bill}")
    print(f"Usage Classification: {usage_classification}")

# Demo
calculate_electricity_bill(250)
```
