Smart Light Controller

```python
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.is_on = False # Light is off by default

    def turn_on(self):
        self.is_on = True
        print(f"{self.name} is ON")

    def turn_off(self):
        self.is_on = False
        print(f"{self.name} is OFF")

    def display_status(self):
        status = "ON" if self.is_on else "OFF"
        print(f"{self.name} is {status}")

# Sample Input
bedroom_light = SmartLight("Bedroom Light")
bedroom_light.turn_on()
bedroom_light.display_status()

living_room_light = SmartLight("Living Room Light")
living_room_light.display_status()
living_room_light.turn_on()
living_room_light.turn_off()
living_room_light.display_status()
```

Employee ID Card System

```python
class EmployeeIDCard:
    def __init__(self, name, employee_id, department):
        self.name = name
        self.employee_id = employee_id
        self.department = department

    def display_id_card(self):
        print("\nEmployee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.employee_id}")
        print(f"Department: {self.department}")

# Sample Input
employee1 = EmployeeIDCard("Rahul", "EMP102", "IT")
employee1.display_id_card()

employee2 = EmployeeIDCard("Priya", "EMP105", "HR")
employee2.display_id_card()
```

Mobile Contact Record

```python
class MobileContact:
    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number
        print("Contact Saved")

    def display_contact_info(self):
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone_number}")

# Sample Input
contact1 = MobileContact("Anita", "9876543210")
contact1.display_contact_info()

print("\n")
contact2 = MobileContact("John Doe", "1234567890")
contact2.display_contact_info()
```

Product Price Tag Generator

```python
class PriceTagGenerator:
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

    def print_price_tag(self):
        print(f"\nProduct: {self.product_name}")
        print(f"Price: ₹{self.price}")

# Sample Input
product1 = PriceTagGenerator("Headphones", 2499)
product1.print_price_tag()

product2 = PriceTagGenerator("Laptop", 65000)
product2.print_price_tag()
```

Movie Rating Display System

```python
class MovieRatingDisplay:
    def __init__(self, movie_name, rating):
        self.movie_name = movie_name
        self.rating = rating

    def display_movie_details(self):
        print(f"\nMovie: {self.movie_name}")
        print(f"Rating: {self.rating} / 5")

# Sample Input
movie1 = MovieRatingDisplay("Inception", 4.8)
movie1.display_movie_details()

movie2 = MovieRatingDisplay("The Matrix", 4.7)
movie2.display_movie_details()
```

Delivery Address Manager

```python
class DeliveryAddressManager:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def print_delivery_details(self):
        print("\nDelivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")

# Sample Input
delivery1 = DeliveryAddressManager("Suman", "Hyderabad")
delivery1.print_delivery_details()

delivery2 = DeliveryAddressManager("Amit", "Bangalore")
delivery2.print_delivery_details()
```
