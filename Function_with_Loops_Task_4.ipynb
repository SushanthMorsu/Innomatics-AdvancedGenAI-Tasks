Mobile Recharge Validation System

```python
# Mobile Recharge Validation System

def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("Recharge amount must be at least ₹50")
        return False

    if amount not in valid_plans:
        print("Invalid plan selected")
        return False

    print("✅ Recharge Successful!")
    print("Selected Plan: ₹", amount)
    return True


# Retry until valid input
while True:
    amount = int(input("Enter recharge plan (199,299,399,599): ₹"))

    if validate_recharge(amount):
        break
    else:
        print("🔄 Invalid input. Please try again.\n")
```

Inventory Reorder Alert System

```python
def check_inventory(products):
    reorder_threshold = 15
    print("\n--- Inventory Reorder Alert System ---")
    for product, stock in products.items():
        if stock < reorder_threshold:
            print(f"Product: {product}, Stock: {stock} - 🚨 Reorder Alert!")
        else:
            print(f"Product: {product}, Stock: {stock} - ✅ Stock OK")

# Sample product inventory
product_stock = {
    "Laptop": 10,
    "Mouse": 50,
    "Keyboard": 12,
    "Monitor": 25,
    "Webcam": 5
}

# Run the inventory check
check_inventory(product_stock)
```

Student Result Processing System

```python
def calculate_result(marks):
    if not marks:
        print("No marks provided.")
        return

    total_marks = sum(marks)
    average_marks = total_marks / len(marks)

    print(f"Student's average marks: {average_marks:.2f}")

    if average_marks >= 50:
        print("Result: Pass")
    else:
        print("Result: Fail")

# Sample usage:
print("--- Student Result Processing System ---")
student1_marks = [75, 80, 60, 90, 70]
print("Student 1:")
calculate_result(student1_marks)

print("\nStudent 2:")
student2_marks = [40, 45, 30, 50, 25]
calculate_result(student2_marks)
```

Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance_km, is_peak_hour):
    base_fare = 50
    rate_per_km = 12
    peak_hour_surcharge_percentage = 0.25 # 25% extra

    # Calculate base fare + distance fare
    fare = base_fare + (distance_km * rate_per_km)

    # Apply peak hour surcharge if applicable
    if is_peak_hour:
        fare *= (1 + peak_hour_surcharge_percentage)

    return fare

# Retry until valid input and user decides to stop
while True:
    try:
        distance_input = input("Enter distance in kilometers: ")
        distance = float(distance_input)

        if distance < 0:
            print("Distance cannot be negative. Please try again.")
            continue

        peak_hour_input = input("Is it peak hour? (yes/no): ").lower()
        is_peak_hour = (peak_hour_input == 'yes')

        estimated_fare = calculate_fare(distance, is_peak_hour)
        print(f"\nEstimated Cab Fare: ₹{estimated_fare:.2f}")

    except ValueError:
        print("Invalid input. Please enter a number for distance.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

    retry = input("\nCalculate another fare? (yes/no): ").lower()
    if retry != 'yes':
        break

print("Thank you for using the Cab Fare Estimator!")
```

Employee Attendance Eligibility Checker

```python
def check_eligibility(attendance_list):
    if not attendance_list:
        return "No attendance records provided."

    present_count = attendance_list.count("P")
    total_days = len(attendance_list)
    attendance_percentage = (present_count / total_days) * 100

    print(f"Attendance Percentage: {attendance_percentage:.2f}%")

    if attendance_percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# Interactive usage:
print("--- Employee Attendance Eligibility Checker ---")

while True:
    user_input = input("Enter attendance records (e.g., PPPAPAA) or 'quit' to exit: ").upper()
    if user_input == 'QUIT':
        break

    # Convert input string to a list of characters
    employee_attendance = list(user_input)

    print("Employee Status:", check_eligibility(employee_attendance))
    print()
```

Password Strength Checker

```python
def check_password_strength(password):
    # Rule 1: Minimum length = 8
    if len(password) < 8:
        print("Password must be at least 8 characters long.")
        return False

    # Rule 2: Must contain at least one digit
    has_digit = False
    for char in password:
        if char.isdigit():
            has_digit = True
            break
    if not has_digit:
        print("Password must contain at least one digit.")
        return False

    # Rule 3: Must contain at least one special character (@#$)
    has_special_char = False
    special_characters = "@#$"
    for char in password:
        if char in special_characters:
            has_special_char = True
            break
    if not has_special_char:
        print("Password must contain at least one special character (@#$).")
        return False

    print("✅ Password Strength: Strong")
    return True

# Interactive usage:
print("--- Password Strength Checker ---")

while True:
    user_password = input("\nEnter a password to check (or 'quit' to exit): ")
    if user_password.lower() == 'quit':
        break
    check_password_strength(user_password)
```
